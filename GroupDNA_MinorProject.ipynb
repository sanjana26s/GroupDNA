```python
def parse_chat(lines):

    messages = []

    for line in lines:

        if " - " not in line:
            continue

        date_time, message_part = line.split(" - ", 1)

        if ": " not in message_part:
            continue

        sender, message = message_part.split(": ", 1)

        messages.append({
            "date_time": date_time,
            "sender": sender,
            "message": message
        })

    return messages
```

```python
from datetime import datetime

def parse_chat(lines):

    messages = []

    for line in lines:

        if " - " in line:

            date_time, message_part = line.split(" - ", 1)

            try:
                datetime.strptime(date_time, "%d/%m/%y, %H:%M")

                if ": " in message_part:
                    sender, message = message_part.split(": ", 1)

                    messages.append({
                        "date_time": date_time,
                        "sender": sender,
                        "message": message
                    })

                else:
                    continue

            except ValueError:
                if messages:
                    messages[-1]["message"] += "\n" + line

        else:
            if messages:
                messages[-1]["message"] += "\n" + line

    return messages
```

```python
messages = parse_chat(lines)

print("Total messages:", len(messages))

for msg in messages[:5]:
    print(msg)
```

```python
members = set()

for msg in messages:
    members.add(msg["sender"])

print("GROUP MEMBERS")
print("--------------------")

for member in sorted(members):
    print(member)

print("--------------------")
print("Total members:", len(members))
```

```python
message_count = {}

for msg in messages:

    sender = msg["sender"]

    if sender in message_count:
        message_count[sender] += 1
    else:
        message_count[sender] = 1

print("MESSAGE COUNT BY MEMBER")
print("------------------------")

for name, count in sorted(message_count.items(),
                          key=lambda x: x[1],
                          reverse=True):
    print(name, ":", count)
```

```python
total_messages = len(messages)

print("GROUPDNA - GROUP OVERVIEW")
print("-------------------------")
print("Total Messages:", total_messages)
print("Total Members:", len(members))
print("Most Active Member:", max(message_count, key=message_count.get))
```

```python
from datetime import datetime

day_count = {}
hour_count = {}

for msg in messages:

    dt = datetime.strptime(
        msg["date_time"], "%d/%m/%y, %H:%M"
    )

    day = dt.strftime("%d/%m/%y")
    hour = dt.hour

    if day in day_count:
        day_count[day] += 1
    else:
        day_count[day] = 1

    if hour in hour_count:
        hour_count[hour] += 1
    else:
        hour_count[hour] = 1

busiest_day = max(day_count, key=day_count.get)
busiest_hour = max(hour_count, key=hour_count.get)

print("GROUP ACTIVITY ANALYSIS")
print("-----------------------")
print("Busiest Day:", busiest_day)
print("Messages on that day:", day_count[busiest_day])
print("Busiest Hour:", busiest_hour)
print("Messages during that hour:", hour_count[busiest_hour])
```

```python
member_names = sorted(members)

activity_matrix = np.zeros((len(member_names), 24), dtype=int)

for msg in messages:

    dt = datetime.strptime(msg["date_time"], "%d/%m/%y, %H:%M")

    sender = msg["sender"]
    hour = dt.hour

    row = member_names.index(sender)

    activity_matrix[row][hour] += 1

print("Activity matrix:")
print(activity_matrix)
```

```python
from collections import Counter
import re

all_words = []

for msg in messages:
    words = re.findall(r'\b[a-zA-Z]+\b', msg["message"].lower())
    all_words.extend(words)

common_words = Counter(all_words)

print("TOP 20 MOST USED WORDS")
print("----------------------")

for word, count in common_words.most_common(20):
    print(word, ":", count)
```

```python
stop_words = {
    "the", "is", "and", "to", "a", "of", "in",
    "it", "you", "i", "for", "on", "this", "that",
    "are", "was", "we", "be", "me", "my", "bro",
    "hai", "ha", "na", "ho", "ko"
}

meaningful_words = [
    word for word in all_words
    if word not in stop_words and len(word) > 2
]

common_words = Counter(meaningful_words)

print("TOP 20 MEANINGFUL WORDS")
print("-----------------------")

for word, count in common_words.most_common(20):
    print(word, ":", count)
```

```python
word_count = {}

for msg in messages:
    sender = msg["sender"]
    words = msg["message"].split()

    if sender not in word_count:
        word_count[sender] = 0

    word_count[sender] += len(words)

print("WORDS USED BY EACH MEMBER")
print("-------------------------")

for name, count in word_count.items():
    print(name, ":", count)
```

```python
print("AVERAGE MESSAGE LENGTH")
print("---------------------")

for name in members:
    total_words = word_count.get(name, 0)
    total_msgs = message_count.get(name, 0)

    average = total_words / total_msgs if total_msgs > 0 else 0

    print(name, ":", round(average, 2), "words per message")
```

```python
from collections import defaultdict
from datetime import datetime

response_times = defaultdict(list)

for i in range(len(messages) - 1):

    current = messages[i]
    next_msg = messages[i + 1]

    if current["sender"] != next_msg["sender"]:

        time1 = datetime.strptime(
            current["date_time"], "%d/%m/%y, %H:%M"
        )

        time2 = datetime.strptime(
            next_msg["date_time"], "%d/%m/%y, %H:%M"
        )

        minutes = (time2 - time1).total_seconds() / 60

        if minutes >= 0:
            response_times[next_msg["sender"]].append(minutes)

print("AVERAGE RESPONSE TIME")
print("---------------------")

for name, times in response_times.items():
    average = sum(times) / len(times)
    print(name, ":", round(average, 2), "minutes")
```

```python
from collections import defaultdict
from datetime import datetime

member_times = defaultdict(list)

for msg in messages:
    dt = datetime.strptime(msg["date_time"], "%d/%m/%y, %H:%M")
    member_times[msg["sender"]].append(dt)

print("LONGEST SILENT PERIOD")
print("---------------------")

for name, times in member_times.items():

    times.sort()

    longest_gap = 0

    for i in range(len(times) - 1):
        gap = (times[i + 1] - times[i]).total_seconds() / 3600

        if gap > longest_gap:
            longest_gap = gap

    print(name, ":", round(longest_gap, 2), "hours")
```

```python
night_messages = {}

for msg in messages:
    dt = datetime.strptime(msg["date_time"], "%d/%m/%y, %H:%M")

    if dt.hour >= 22 or dt.hour < 5:
        sender = msg["sender"]

        if sender not in night_messages:
            night_messages[sender] = 0

        night_messages[sender] += 1

print("NIGHT OWL ANALYSIS")
print("------------------")

for name, count in sorted(
    night_messages.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(name, ":", count, "night messages")
```

```python
print("GROUP SPAMMER ANALYSIS")
print("----------------------")

spammer = max(message_count, key=message_count.get)

print("Most frequent sender:", spammer)
print("Messages sent:", message_count[spammer])
```

```python
question_count = {}

for msg in messages:
    if "?" in msg["message"]:
        sender = msg["sender"]

        if sender not in question_count:
            question_count[sender] = 0

        question_count[sender] += 1

print("QUESTION MASTER ANALYSIS")
print("------------------------")

if question_count:
    question_master = max(question_count, key=question_count.get)
    print("Question Master:", question_master)
    print("Questions asked:", question_count[question_master])
else:
    print("No questions found.")
```

```python
storyteller = max(
    members,
    key=lambda name: word_count.get(name, 0) / message_count.get(name, 1)
)

print("STORYTELLER ANALYSIS")
print("--------------------")
print("Storyteller:", storyteller)
print("Average words per message:",
      round(word_count[storyteller] / message_count[storyteller], 2))
```

```python
laughter_words = [
    "haha", "hahaha", "lol", "lmao", "😂", "🤣"
]

comedy_count = {}

for msg in messages:
    text = msg["message"].lower()

    for word in laughter_words:
        if word in text:
            sender = msg["sender"]
            comedy_count[sender] = comedy_count.get(sender, 0) + 1
            break

print("COMEDIAN ANALYSIS")
print("-----------------")

if comedy_count:
    comedian = max(comedy_count, key=comedy_count.get)
    print("Group Comedian:", comedian)
    print("Funny messages:", comedy_count[comedian])
else:
    print("No laughter-related messages found.")
```

```python
ghost = min(message_count, key=message_count.get)

print("GHOST MEMBER ANALYSIS")
print("---------------------")
print("Least active member:", ghost)
print("Messages sent:", message_count[ghost])
```

```python
caring_words = [
    "take care", "eat", "sleep", "food",
    "health", "safe", "help", "study"
]

care_count = {}

for msg in messages:
    text = msg["message"].lower()

    for word in caring_words:
        if word in text:
            sender = msg["sender"]
            care_count[sender] = care_count.get(sender, 0) + 1

print("GROUP MOM ANALYSIS")
print("------------------")

if care_count:
    group_mom = max(care_count, key=care_count.get)
    print("Group Mom:", group_mom)
    print("Caring messages:", care_count[group_mom])
else:
    print("No caring words found.")
```

```python
emotion_words = [
    " omg", "wow", "seriously", "shit",
    "damn", "angry", "sad", "cry",
    "unbelievable", "hate", "love"
]

drama_count = {}

for msg in messages:
    text = " " + msg["message"].lower()

    for word in emotion_words:
        if word in text:
            sender = msg["sender"]
            drama_count[sender] = drama_count.get(sender, 0) + 1

print("DRAMA QUEEN ANALYSIS")
print("--------------------")

if drama_count:
    drama_queen = max(drama_count, key=drama_count.get)
    print("Drama Queen:", drama_queen)
    print("Emotional messages:", drama_count[drama_queen])
else:
    print("No emotional keywords found.")
```

```python
personality_results = {
    "Spammer": spammer,
    "Group Mom": group_mom if care_count else "Not identified",
    "Night Owl": max(night_messages, key=night_messages.get)
                 if night_messages else "Not identified",
    "Storyteller": storyteller,
    "Drama Queen": drama_queen if drama_count else "Not identified",
    "Ghost": ghost,
    "Comedian": comedian if comedy_count else "Not identified",
    "Question Master": question_master
                     if question_count else "Not identified"
}

print("GROUPDNA - PERSONALITY REPORT")
print("-----------------------------")

for personality, member in personality_results.items():
    print(personality, ":", member)
```
